#  Task 03:
The marketing team wants to evaluate if their marketing spend is actually helping to boost sales. 

Plot **Marketing Spend vs Units Sold** and see whether spending more money always increases sales across different customer segments.

```python
import matplotlib.pyplot as plt

color_map = {
    "Occasional": "red",
    "Premium": "blue",
    "Regular": "green"
}


plt.figure(figsize=(10,6))

for seg in df["Customer_Segment"].unique():  # this will look for every unique occurance to create a subset of it . 
    subset = df[df["Customer_Segment"] == seg]
    plt.scatter(subset["Marketing_Spend"], subset["Units_Sold"], color=color_map[seg], alpha=0.6, label=seg)

plt.title("Marketing Spend vs Units Sold by Customer Segment", fontsize=12, fontweight="bold")
plt.xlabel("Marketing Spend ($)", fontsize=12, fontweight="bold")
plt.ylabel("Units Sold", fontsize=12, fontweight="bold")
plt.legend(title="Customer Segment", fontsize=6, edgecolor="black")
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```

- Looking at this, we can clearly see the data points are scattered in the middle.
- for none of the category there is a trend of that increasing marketing spend always increases units sold across customer segments.
- We can definitly say that *****increasing marketing cost isnt favoruing their business at all because number of units is not increasing.*****

# Task 04:
The operations team wants to understand typical order sizes. 

Plot Units Sold to check whether most purchases are small (few units) or if bulk orders are more common.

```python
units = df['Units_Sold']

mean_val = units.mean()
median_val = units.median()
mode_val = units.mode()[0]
standard_dev   = units.std()

print(f"Mean of  Units Sold: {mean_val:.2f}")
print(f"Median of Units Sold: {median_val}")
print(f"Mode of Units Sold: {mode_val}")
print("Standard Deviation of  Units Sold:", standard_dev)

```

-  It verifies our visual data that most most repeated no of units that are ordered is 30, the averaeg is also arouund 29 and the most common sales are 7 units around mean like 29+ 7.27 = 36.27 and 29-7.27 = 21.73. = 22
- Which mean most ordered/sales are placed for 22 to 36 units.

```python

avg_units = df['Units_Sold'].mean() # mean 
standard_dev = df['Units_Sold'].std() # standard deviation


plt.figure(figsize=(10,6)) # Plotting histogram of Units Sold
plt.hist(df['Units_Sold'], bins=15, color='skyblue', edgecolor='black', alpha=0.7)

# Highlight the background behind the most common orders
# This shades the area between -1 and +1 standard deviation
plt.axvspan(avg_units - standard_dev, avg_units + standard_dev, color='salmon', alpha=0.3, label='Common Units Orders')

#here we are highlighting the background: starting from ( 22, upto  36, highlighting color is red, with minimal shading and label

plt.axvline(avg_units - standard_dev, color='grey', linestyle='--', linewidth=1.5) # drawing std line from left side
plt.axvline(avg_units + standard_dev, color='grey', linestyle='--', linewidth=1.5) # drawing std line friom right


plt.xticks(np.arange(0, df['Units_Sold'].max()+5, 5)) # values are 5 point apart on x axis,
plt.axvline(avg_units, color='black', linestyle='-', linewidth=1.5, label=f'Average = {avg_units:.1f}') # Adding a vertical line for average units sold
plt.legend()

plt.title("Distribution of Units Sold per Order", fontsize=12, fontweight="bold")
plt.xlabel("Units Sold", fontsize=12, fontweight="bold")
plt.ylabel("Number Of Orders", fontsize=12, fontweight="bold")
plt.grid(True)

plt.show()
```

- here we have highlighted the area of the 1st standard deviation where most of the ordered units lies


## Conclsuion:

- from this graph its clear that the most populated bins or the bins having higher count is in the range of 22 - 36 . 
- we can also see it from the average, that the high frequency bins are around it which means that the most orders placed are typpically of size 22 to 36.
- The data is mostly clumped in the middle, and most orders are for a medium-sized number of units, not a very small or very large quantity.
- The average order size is about 30 units, which is a good representation of what most customers are buying.
- Overall, it shows that distribution is normal which means *****neither small nor too bulk order are preferred by customer rather medium sized orders are the most prefereed one*****

#### This means that:
- Customers trust the store enough to buy a few items at a time, rather than just one or two.
- This is also a  good thing for the operations team because it means they are handling a consistent and predictable number of items per order, which makes things like packaging and shipping much more efficient.
- The store is likely not a place for customers looking to buy a single low-priced item, and it's also not a wholesale shop for people buying hundreds of units at a time. It's somewhere in the middle.

# Task 05:
The sales department is curious about the company’s discount strategy. 

Create a Plot of **Discount values** to understand whether the company usually gives small discounts, moderate ones, or very high discounts to customers.ts.

```python
avg_d = df['Discount'].mean() # mean 

plt.figure(figsize=(13,8)) # Plotting histogram of Units Sold
plt.hist(df['Discount'], bins=25, color='teal', edgecolor='black', alpha=0.7)

#here we are highlighting the background: starting from ( 22, upto  36, highlighting color is red, with minimal shading and label

plt.xticks(np.arange(0, df['Discount'].max()+5, 5)) # values are 5 point apart on x axis,
plt.axvline(avg_d, color='green', linestyle='-', linewidth=2, label=f'Average = {avg_d:.1f}') # Adding a vertical line for average units sold
plt.legend(fontsize=10)

plt.title("Distribution of Discount per Order", fontsize=12, fontweight="bold")
plt.xlabel("Discount Percentage (%)", fontsize=12, fontweight="bold")
plt.ylabel("Frequency of Orders", fontsize=12, fontweight="bold")
plt.grid(True)

plt.show()
```

## Conclusion:

- The distribution of discounts per order shows a relatively flat across the entire range, indicating that the company gives a variety of discount levels rather than focusing on a single, narrow band.
- The histogram shows a wide spread of discount percentages from 0% to 50%.
- The average discount is approximately 24.9%, as indicated by the green vertical line ( which we drew for understanding that bins distribution on either side of mean is almost equal )
- The most frequent discounts appear to be in the 15-20% and 30-35% ranges, although the differences between the bars are not high.
- There are no significant peaks or valleys, suggesting that all discount levels are given frequently.
- *****Overall, the company employs a diverse discount strategy that isn't heavily skewed towards either very small or very large discounts.*****
##### Thiss suggest that:

- Small Discounts (0-10%):mmight be used for promotions like "first-time buyer" or for popular, high-margin products that don't need a deep discount to sell.

- Moderate Discounts (15-35%): are  the most common range, suggesting it's the standard for seasonal sales, loyalty program rewards, or clearing out older inventory. The average discount of 24.9% falls in this category.

- High Discounts (40-50%): llikely for specific, strategic purposes, such as getting rid of end-of-season stock, for special flash sales, or for very high-value customers as a special reward.
