## Task 1: Best-Selling Product Category

Your manager wants to know which product category generated the highest total sales amount. 

Create a bar chart showing total sales (Y-axis) for each product category (X-axis). Highlight which category is the best seller."

```python
import matplotlib.pyplot as plt

category_sales = df.groupby("Product Category")["Total Amount"].sum()  # first we are calculating total amount with respect to each cetegory and then summing them up.
bars = plt.bar(category_sales.index, category_sales.values, color="skyblue") 


best_category = category_sales.idxmax()  # now we are getting the label of the category that has the highest sales 
best_value = category_sales.max()  # here we are getting the sales value ( numerical value ) of category that had highest sales


for i in range(len(bars)):  # if there are 3 bars then go through all three.
    if category_sales.index[i] == best_category: # if the index (means the category name because its 0th and 1st is the values of amount) of 
                                                  # iterating bar matchhes with the best category name 
        bars[i].set_color("orange")              # set it color to orrange

    
for i, v in enumerate(category_sales.values):   # here we are writing the total sales amount of each product category obver its bar  
    plt.text(i, v + 50, str(int(v)), ha='center') # i is the index of bar ( 0,1,2 ) v is the sales amount.

#plt.text: put text over bar, v+50 put the text 50 points above bar so the text is clean and not messy woth bar. ha means its in center.

plt.xlabel("Product Category")   # setting x labell
plt.ylabel("Total Sales Amount")   # here we are setting y axis label
plt.title("Best-Selling Product Category")  # here we are setting the title
plt.show()  # finally we are showing the graph.

print("Best selling category product is :", best_category)
print( "The total sales amount of is: ", best_value)
```

## Explanation : 
- We first imported the library to make our graphs.

## Task 4: Age Group Comparison

To understand customer demographics, divide customers into 3 groups:

Young (18–30)

Adult (31–50)

Senior (51+)

Make a bar chart showing total sales amount by these age groups. Which age group spends the most?"

```python
import numpy as np

df["AgeGroup"] = np.where((df["Age"] >= 18) & (df["Age"] <= 31), "Young (18-31)",
                  np.where((df["Age"] >= 32) & (df["Age"] <= 50), "Adult (32-50)",
                           "Senior (51+)"))

sales_by_age= df.groupby("AgeGroup")["Total Amount"].sum()
print(sales_by_age.idxmax())
print(sales_by_age.max())
```

## Explanation
- Since our Age column doesn’t have intervals, we need to divide it into groups.
- Using numpy with np.where, we assign each age into an interval.
- The first condition says:
- If age is greater than or equal to ( 18 and less than or equal to 31 ) : label it as Young (18-31).
- The same logic is applied for the next two intervals: 32–50:  Adult,   51+:  Senior
- Now that we have assigned our age column labels based on requiremnets, we can apply group by on AgeGroup column and Total amount column to get total sales by these age groups.

```python


bars=plt.bar(sales_by_age.index, sales_by_age.values, color="skyblue", edgecolor="green")

# to find ehich age group spends the most
max_spend_name= sales_by_age.idxmax()


for i in range(len(bars)): 
    if sales_by_age.index[i] == max_spend_name:  # finding which group spends the most by iterating through all
     
        bars[i].set_color("orange")
        bars[i].set_edgecolor("green")
      
for i, v in enumerate(sales_by_age.values):   # like all above we are wriitng the total values here
    plt.text(i, v + 50, str(int(v)), ha='center')

plt.xlabel("Age Group", fontweight="bold")
plt.ylabel("Total Spending / sales amount", fontweight="bold")
plt.title("Total Spending by Age Group", fontsize=14, fontweight="bold")
plt.xticks(fontweight="bold")
plt.show()

print("The age group that spends the most is:", max_spend_name)

```

## Explanation:
- After applying group by, the rest of the method is same to create bars, highlighting the category bar that had the most purchases or contributed to sales, giving bars edge, setting the x and y labels/ titles,and giving whole graph title.
- We have also shown the contribution to sales amount by each age group over the top of its bar. The method of making this graph is almost same as above all except that we had to provide labels to age group so we can perform group by easily.

## Task 3: Monthly Sales Trend (Bar Version)

The finance team wants to check monthly performance. 

Convert the Date column into months and make a bar chart of total sales amount per month. Identify which month had the highest sales.

```python
df.dtypes
```

```python

df["Date"] = pd.to_datetime(df["Date"]) # converting it into dattime datatype

df["Month"] = df["Date"].dt.month_name() # now we are extracting the months from all dates and storing it into separate tcolumn apart from date column

monthly_sales = df.groupby("Month")["Total Amount"].sum().sort_values() 
print(monthly_sales)

```

```python
df.dtypes
```

## Explanation:
- From above we know that date column does not have datetime datatype.So, we first converted it into date time dataype by usingg pandas.
- Then after storing the months into separate column we need to find the highest sales in a month which is to be done by grouping the months by total amount.
- We have also sorted our results, so when we use varaible in plotting, the bars are in sequence and good.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))
bars= plt.bar(monthly_sales.index, monthly_sales.values, color="skyblue") # creating bars heer like previously

best_month = monthly_sales.idxmax()  # here we are getting the name of the month having the highest sales.
best_value = monthly_sales.max()  # here the total sales number the month had

for i in range(len(bars)):   # itertsaing through all the bars of months
    if monthly_sales.index[i]== best_month:  # if current iterating month name is same as the highest sales month 
        bars[i].set_color("pink")     # then change its colors and give it a border to highlight
        bars[i].set_edgecolor("purple")


   
for i, v in enumerate(monthly_sales.values):   # here we are writing the total sales amount of each product category obver its bar  
    plt.text(i, v + 50, str(int(v)), ha='center') # i is the index of bar ( 0,1,2 ) v is the sales amount.

    
plt.title("Total Sales Amount per Month", fontsize=14, fontweight="bold")   # giving graph a title
plt.xlabel("Month", fontsize=12, fontweight="bold")
plt.ylabel("Total Sales", fontsize=12, fontweight="bold")
plt.xticks(rotation=45, fontweight="bold")
plt.show()
```
