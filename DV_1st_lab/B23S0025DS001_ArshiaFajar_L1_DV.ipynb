```python
import numpy as np  # importing the numpy library
```

#### 1. 10 points generation 

- Firts we imported our numpy library.
- Linspace ( start, stop , number ) : where start is the number from where we want the generation to start, stop is the number till where we want out numbers to be generated, number is the total number of poinst we want to generate.
- For then points generation between 0 and 10 it would be linspace (0,10,10 ) :
- start from 0
- Go till 10
- create 10 points equally distanced.

#### Output

```python
data_of_10= np.linspace(0,10,10)
print("Ten data points between 0 and 10 are: \n" , data_of_10)
```

- If we wantt to see what exactly is the difference between each consecutive point we can use .diff keyword as given.
- This will give us the difference between each point .

```python
distance_10 = np.diff(data_of_10)
print(distance_10)
```

- Now we can see that each number has a difference of 1.1111 ( eg 0 + 1.1111  = 1.1111, 1.1111+1.1111 = 2.2222 etc) 

#### 2. 100 points generation

- Just like we did for 10 we will do the same for this one except that the last parameter in linspace woudl be 100.
- linspace ( 0,10,100 ) meaning we are creating total of 100 points between 0 and 10.

#### Output

```python
data_of_100 = np.linspace ( 0,10,100)
print("100 data points between 0 and 10 are: \n" , data_of_100)
```

- Now we can see that difference between consective points has reduced since we have to get 100 points so each point is only points away from each other. ( in decimals unlike previous one which was in whole number form)

```python
distance_100 = np.diff(data_of_100)
print(distance_100[: 5])
```

#### 3. 200 Points generation 
- The last parameter in linspace would be 200 as given in the code below

#### Output

```python
data_of_200=np.linspace(0,10,200)
print("Two hundred data points between 0 and 10 are: \n" , data_of_200)
```

- From here we can see that the difference has been reduced more since we have get more points between 0 and 10. Therefore every points is only 0.05 points away from next point .

```python
distance_200 = np.diff(data_of_200)
print(distance_200 [:5])
```

## Task 2  Function Application 
-  Use the provided functions to compute output values from the generated dataset.𝑥 +1

- f(x)= (sin (x) + cos (2x) )  ( e ^ -x) 
- f(x) = (sin (x^2) . cos (3x) + ln ( x+1) )) . e^(-0.5x) + ( √ x + 1)

#### Functions 
- First we coded the fuunctions:
- exp: means exponential
- ** means power of.
- sqrt means square root.

```python
def fun1(x): # this is for the function1 
    return (np.sin(x) + np.cos(2*x)) * np.exp(-x)

def fun2(x): # this iis for the function 2 
    return (np.sin(x**2) * np.cos(3*x) + np.log(x+1)) * np.exp(-0.5*x) + np.sqrt(x+1)
```

#### Functions on 10 data points

- from the last implementation we have saved the result in the variables.
- Now to implement the 1st function we have passed the variable ( data_of_10) to 1st function.
- The result will be returned from the function and saved into variable through which we will printthe result. 

#### 1st function

```python
function1_on_10= fun1(data_of_10)
print(function1_on_10)
```

#### 2nd Function

- Like the previous function we have implemented the seconf function on 10 points.

```python
function2_on_10= fun2(data_of_10)
print(function2_on_10)
```

#### Functions on 100 data points

#### 2nd Function

- We have done the same and passed the points variable in 2nd function.

#### Functions on 200 data points 

#### 1st Function

- Like the above, we have done the same for the 200 data points by passing the variable into function 1 and function 2

## Task 3  Visualization 

#### Visualization of Function 1

- First we defined the size of graph .
- Now to make a line plot we need values on both axis x and y.
- On x axis, we have set our own generated 10, 100 and 200 data points stored in varaibles " data_of_10", "data_of_100", "data_of_200" respevtively.
- On y axis we have placed the values we got after applying function 1 on all three sets on data points.
- We have then set the title, label for x and y axis.
- The legend is then set which will show the color of line representing trend of each data points.
- Finally we have shown our graph.


```python
import matplotlib.pyplot as plt  # importing the library
```

```python

plt.figure(figsize=(10,6))
plt.plot(data_of_10, function1_on_10, label="Function 1 on 10 data points")
plt.plot(data_of_100, function1_on_100, label="Function 1 on 100 data points")
plt.plot(data_of_200, function1_on_200, label="Function 1 on 200 points")
plt.title("Function 1 comparison on 10,100 and 200 data points")
plt.xlabel("Generated Data Points")
plt.ylabel("Results after Function 1")
plt.legend()
plt.grid(True)
plt.show()


```

- if we look at the above graph we can see that the blue line (10 points ) show no peaks. It just a line with no obvious peaks.
- However, if we look at the green ( 200 points ) and orange line (100 points) which is following the path of green line has more peaks. It is beacuse these 2 had more points to connect as compare to 10 data points

#### Visualization of Function 2

```python
plt.figure(figsize=(10,6))
plt.plot(data_of_10, function2_on_10, label="Function 2 on 10 data points")
plt.plot(data_of_100, function2_on_100, label="Function 2 on 100 data points")
plt.plot(data_of_200, function2_on_200, label="Function 2 on 200 data  point")
plt.title("Function 2 on 10,100 and 200 data points")
plt.xlabel("Generated Data Points")
plt.ylabel("Results after Function 2")
plt.legend()
plt.grid(True)
plt.show()
```

- In this we can see that 200 points forms the more smoother graph then the 10 data points which is a simple line instead of any peaks.
- This shows that the more the data points, the more smoother shape of the graph will be.
